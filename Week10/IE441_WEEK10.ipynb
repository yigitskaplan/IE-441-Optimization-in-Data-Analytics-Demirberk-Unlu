```python
import pandas as pd
from sklearn import datasets
```

```python
# hazır datasetler
```

```python
cancer = datasets.load_breast_cancer()
```

```python
print(cancer.feature_names)
```

```python
# features = x'ler
```

```python
print(cancer.target_names)
```

```python
#targer = y'ler
```

```python
df = pd.DataFrame(cancer.data, columns = cancer.feature_names)
```

```python
df.head()
```

```python
df['target'] = cancer.target
```

```python
df.head()
```

```python
# ['malignant' = 0, 'benign' = 1] olarak, target columunun altına ekleniyor
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df.isnull().sum()
```

```python
# no null values
```

```python
df['target'].value_counts()
```

```python
# 212 malignant, 357 benign
```

```python
sns.pairplot(df,hue='target', palette=['red','green'])
```

```python
# our goal is to seperate red and green values
```

```python
X = df.drop(['target'], axis = 1)
Y = df['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, train_size = 0.8, stratify = Y,random_state = 149)
```

```python
svc = svm.SVC()
```

```python
svc.fit(x_train,y_train)
```

```python
df['target'].value_counts()/len(df)
```

```python
# If its keep guessing 1, its accuracy would be 63%. Our algorithm needs to get better score than 63%
```

```python
y_pred = svc.predict(x_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
# Our score for the "kernel = radial basis, c = 1" is 87%, its better than 63%. Acceptable
```

```python
svc_c10 = svm.SVC(C=10)
```

```python
# kernel = radial basis, c = 10
```

```python
svc_c10.fit(x_train, y_train)
```

```python
y_predc10 = svc_c10.predict(x_test)
```

```python
accuracy_score(y_test, y_predc10)
```

```python
# its better than, radial basis, c = 1. Thus when c is larger than 1 to the some point it would give better accuracy.
```

```python
svc_c10_lin = svm.SVC(kernel = 'linear', C=10)
```

```python
svc_c10_lin.fit(x_train,y_train)
```

```python
y_predc10lin = svc_c10_lin.predict(x_test)
```

```python
accuracy_score(y_test, y_predc10lin)
```

```python
# its better than radial, c = 10. Thus kernel = liner better algorithm than radial in our case.
```

```python
svc_c10_sig = svm.SVC(kernel='sigmoid', C=10)
```

```python
svc_c10_sig.fit(x_train,y_train)
```

```python
y_predc10sig = svc_c10_sig.predict(x_test)
```

```python
accuracy_score(y_test, y_predc10sig)
```

```python
# kernel = sigmoid worse than linear, linear kernel still optimal
```

```python
svc_c10_poly = svm.SVC(kernel = 'poly', C=10)
```

```python
svc_c10_poly.fit(x_train, y_train)
```

```python
y_predc10poly = svc_c10_poly.predict(x_test)
```

```python
accuracy_score(y_test, y_predc10poly)
```

```python
# our algorithms kernels: linear>radial>poly>sigmoid, C when bigger than 1 we get a more accurate results.
# highest accuracy out of all our tries : kernel = 'linear', C=10 with 0.947 accuracy
```
