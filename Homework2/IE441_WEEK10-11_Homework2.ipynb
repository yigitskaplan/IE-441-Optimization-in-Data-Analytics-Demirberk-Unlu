```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(data = iris.data, columns = iris.feature_names)
```

```python
df['target'] = iris.target
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
# aynı oldukları için stratify yapmaya gerek yok
```

```python
sns.pairplot(df, hue='target', palette = ['magenta', 'green', 'blue'])
```

```python
X = df.drop(['target'], axis = 1)
Y = df['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, train_size=0.8, random_state = 2024)
```

```python
dt = DecisionTreeClassifier(random_state=0)
```

```python
dt.fit(x_train,y_train)
```

```python
plot_tree(dt)
```

```python
plt.figure(figsize=(12,12))
plot_tree(dt,fontsize = 10)
```

```python
y_train_pred = dt.predict(x_train)
```

```python
accuracy_score(y_train, y_train_pred)
```

```python
y_test_pred = dt.predict(x_test)
```

```python
accuracy_score(y_test, y_test_pred)
```

```python
# ----------------------------------------------------
```

```python
dt2 = DecisionTreeClassifier(criterion = 'entropy', max_depth=2, random_state=0)
```

```python
dt2.fit(x_train,y_train)
```

```python
plt.figure(figsize=(12,12))
plot_tree(dt2)
```
