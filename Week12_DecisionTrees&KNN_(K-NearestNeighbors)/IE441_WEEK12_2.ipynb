```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('Iris.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
df.drop(['Id'], axis = 1, inplace = True)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Species'].value_counts()
```

```python
sns.pairplot(df, hue = 'Species', palette = ['red', 'blue', 'green'])
```

```python
X = df.drop(['Species'], axis = 1)
Y = df['Species']
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state= 0)
```

```python
knn = KNeighborsClassifier()
```

```python
knn.fit(x_train, y_train)
```

```python
y_hat_train = knn.predict(x_train)
```

```python
accuracy_score(y_hat_train,y_train)
```

```python
y_hat_test = knn.predict(x_test)
```

```python
accuracy_score(y_hat_test, y_test)
```

```python
knn_k10 = KNeighborsClassifier(n_neighbors=10)
```

```python
knn_k10.fit(x_train, y_train)
```

```python
y_hat_train_10 = knn_k10.predict(x_train)
```

```python
accuracy_score(y_hat_train_10, y_train)
```

```python
y_hat_test_10 = knn_k10.predict(x_test)
```

```python
accuracy_score(y_hat_test_10, y_test)
```
